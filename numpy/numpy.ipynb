```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset('tips')
```

```python
# heat map
```

```python
pivot = df.pivot_table(
    index='day',
    columns='time',
    values='total_bill',
    aggfunc='mean'
)

sns.heatmap(pivot)
plt.show()
```

```python
#pair plot
sns.pairplot(df)
plt.show()
```

```python
#Pair Plot 
sns.pairplot(df, hue = 'sex' ,kind = 'kde')
plt.show()
```

```python
#strip Plot

sns.stripplot(data = df, x = 'day' , y = 'total_bill', hue = 'sex')
plt.show
```

```python
#Cat Plot
```

```python
sns.catplot(data = df , x = 'day', y = 'total_bill' ,kind='bar', hue = 'sex') 
plt.show
```
